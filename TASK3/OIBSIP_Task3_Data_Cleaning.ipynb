<h3>📞 Phone Number Formatting</h3>

<p>
The <strong>Phone</strong> column was initially stored as a numeric data type and contained negative values. Since phone numbers are identifiers rather than numerical quantities, the column was converted to text format and negative signs were removed.

After applying these transformations, the updated data types and values were verified to ensure successful cleaning.

```python
# ==========================================
# FIX DATE AND PHONE DATA
# ==========================================

# Convert Join_Date to datetime format
df['Join_Date'] = pd.to_datetime(
    df['Join_Date'],
    errors='coerce'
)

# Convert Phone to string and remove negative sign
df['Phone'] = (
    df['Phone']
    .astype(str)
    .str.replace('-', '', regex=False)
)

# Check results
print("Join_Date Data Type:", df['Join_Date'].dtype)
print("Missing Join_Date Values:", df['Join_Date'].isnull().sum())

print("\nPhone Data Type:", df['Phone'].dtype)
print("\nPhone Sample:")
print(df['Phone'].head())
```

<h2>📊 Outlier Detection Using the IQR Method</h2>

<p>
Outlier detection was performed on the numerical columns <strong>Age</strong> and <strong>Salary</strong> using the Interquartile Range (IQR) method.
</p>

<p>
The IQR method identifies potential outliers by calculating the first quartile (Q1), third quartile (Q3), and the interquartile range.
</p>

<p>
Values falling below the lower bound or above the upper bound are considered potential outliers.
</p>

<h3>📐 IQR Formula</h3>

<p>
<strong>IQR = Q3 − Q1</strong>
</p>

<p>
<strong>Lower Bound = Q1 − 1.5 × IQR</strong><br>
<strong>Upper Bound = Q3 + 1.5 × IQR</strong>
</p>

```python
# ==========================================
# OUTLIER DETECTION USING IQR
# ==========================================

numerical_columns = ['Age', 'Salary']

for column in numerical_columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]

    print(f"\n{'=' * 50}")
    print(f"Column: {column}")
    print(f"Q1: {Q1}")
    print(f"Q3: {Q3}")
    print(f"IQR: {IQR}")
    print(f"Lower Bound: {lower_bound}")
    print(f"Upper Bound: {upper_bound}")
    print(f"Number of Outliers: {len(outliers)}")
```

<h3>🔍 Outlier Detection Results</h3>

<ul>
  <li><strong>Age:</strong> No outliers were detected.</li>
  <li><strong>Salary:</strong> No outliers were detected.</li>
</ul>

<p>
Based on the IQR method, no values required removal or modification. Therefore, all valid records were retained in the dataset.
</p>
